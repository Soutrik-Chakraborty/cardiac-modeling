```python
import os
import argparse
import zipfile
import tempfile
import shutil
from pathlib import Path
import numpy as np
import pandas as pd
import nibabel as nib
from scipy import ndimage
from skimage import measure
import trimesh
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
f1_score, roc_auc_score, confusion_matrix)
import joblib
```

```python
def read_metadata(excel_path):
    df = pd.read_csv(excel_path)
    return df
```

```python
def find_image_and_mask_in_zip(zip_path, patient_prefix):
  with zipfile.ZipFile(zip_path, 'r') as zf:
    names = zf.namelist()
    # simple heuristics
    img = None
    mask = None
    for n in names:
      bn = Path(n).name.lower()
      if bn.startswith(patient_prefix) and bn.endswith(('.nii', '.nii.gz')):
        if '_seg' in bn or 'seg' in bn or 'label' in bn:
          if mask is None:
            mask = n
      else:
        if img is None:
          img = n
      return img, mask
```

```python
def load_nifti_from_zip(zip_path, member_name):
  with zipfile.ZipFile(zip_path, 'r') as zf:
    with tempfile.TemporaryDirectory() as td:
      zf.extract(member_name, td)
      extracted = Path(td) / member_name
# If nested directories in name, ensure correct path
      if not extracted.exists():
# join the tail
        extracted = Path(td) / Path(member_name).name
      img = nib.load(str(extracted))
      arr = img.get_fdata()
# nibabel returns zooms as (x,y,z) typically
      zooms = img.header.get_zooms()
# Convert to spacing order (z,y,x)
      spacing = (zooms[2], zooms[1], zooms[0]) if len(zooms) >= 3 else (1.0,1.0,1.0)
      return arr, spacing
```

```python
def combine_labels_to_binary(mask_arr, keep_labels=None):
  if keep_labels is None:
    return (mask_arr > 0).astype(np.uint8)
  out = np.zeros_like(mask_arr, dtype=np.uint8)
  for l in keep_labels:
    out = np.logical_or(out, mask_arr == l)
  return out.astype(np.uint8)
```

```python
def expand_mask_by_mm(mask, spacing, expand_mm):
  mask_bool = mask.astype(bool)
  inv = ~mask_bool
  dt_mm = ndimage.distance_transform_edt(inv, sampling=spacing)
  expanded = (dt_mm <= expand_mm) | mask_bool
  return expanded.astype(np.uint8)
```

```python
def dice_score(a, b):
  a = a.astype(bool)
  b = b.astype(bool)
  if a.sum() + b.sum() == 0:
    return 1.0
  return 2.0 * (a & b).sum() / (a.sum() + b.sum())
```

```python
def iou_score(a, b):
  a = a.astype(bool)
  b = b.astype(bool)
  inter = (a & b).sum()
  union = (a | b).sum()
  if union == 0:
    return 1.0
  return inter / union
```

```python
def hausdorff_distance(a, b, spacing):
# compute Hausdorff (symmetric) using surface points and euclidean distances
  from scipy.spatial.distance import cdist
  a_pts = np.argwhere(a.astype(bool)) * np.array(spacing)
  b_pts = np.argwhere(b.astype(bool)) * np.array(spacing)
  if a_pts.size == 0 or b_pts.size == 0:
    return np.nan
  d_ab = cdist(a_pts, b_pts)
  d_ba = cdist(b_pts, a_pts)
  return max(d_ab.min(axis=1).max(), d_ba.min(axis=1).max())
```

```python
def mesh_from_mask(mask, spacing, out_mesh_path):
# marching_cubes expects mask as float and spacing tuple (z,y,x)
  verts, faces, normals, values = measure.marching_cubes(mask.astype(float), level=0.5, spacing=spacing)
  mesh = trimesh.Trimesh(vertices=verts, faces=faces, vertex_normals=normals, process=False)
  mesh.export(out_mesh_path)
  return out_mesh_path
```

```python
# ---------------------- Feature extraction ----------------------

def extract_geometric_features(mask, spacing):
    # basic features: volume (mm3), centroid, bbox sizes, surface area (approx from mesh)
    voxel_vol = np.prod(spacing)
    vol_mm3 = mask.sum() * voxel_vol
    # centroid in voxel coords -> mm
    pts = np.argwhere(mask.astype(bool))
    if pts.size == 0:
        return {
            'volume_mm3': 0,
            'centroid_x_mm': np.nan,
            'centroid_y_mm': np.nan,
            'centroid_z_mm': np.nan,
            'bbox_x_mm': np.nan,
            'bbox_y_mm': np.nan,
            'bbox_z_mm': np.nan,
            'surface_area_mm2': np.nan,
            'compactness': np.nan,
        }
    centroid_vox = pts.mean(axis=0)
    centroid_mm = centroid_vox * np.array(spacing)
    minp = pts.min(axis=0) * np.array(spacing)
    maxp = pts.max(axis=0) * np.array(spacing)
    bbox = maxp - minp
    # surface area via marching cubes
    verts, faces, normals, _ = measure.marching_cubes(mask.astype(float), level=0.5, spacing=spacing)
    mesh = trimesh.Trimesh(vertices=verts, faces=faces, process=False)
    surf_area = mesh.area
    # compactness: (surface_area^1.5)/volume or similar metric
    compactness = (surf_area ** 1.5) / (vol_mm3 + 1e-9)
    return {
        'volume_mm3': float(vol_mm3),
        'centroid_x_mm': float(centroid_mm[2]),  # careful ordering (z,y,x) -> we map to x,y,z as desired
        'centroid_y_mm': float(centroid_mm[1]),
        'centroid_z_mm': float(centroid_mm[0]),
        'bbox_x_mm': float(bbox[2]),
        'bbox_y_mm': float(bbox[1]),
        'bbox_z_mm': float(bbox[0]),
        'surface_area_mm2': float(surf_area),
        'compactness': float(compactness),
    }
```

```python
# ---------------------- Average model ----------------------

def build_average_model(masks_list, spacing, out_path_prob, out_path_mesh, thresh=0.5):
    # masks_list: list of binary numpy arrays with same shape & spacing
    mean_prob = np.mean(np.stack(masks_list, axis=0), axis=0)
    # save average probability map as NIfTI
    affine = np.diag([spacing[2], spacing[1], spacing[0], 1.0])
    nib.save(nib.Nifti1Image(mean_prob.astype(np.float32), affine), out_path_prob)
    # threshold to binary and mesh
    avg_mask = (mean_prob >= thresh).astype(np.uint8)
    mesh_from_mask(avg_mask, spacing, out_path_mesh)
    return mean_prob, avg_mask
```

```python
# ---------------------- Classifier ----------------------

def train_chd_classifier(X, y, out_model_path):
    # simple pipeline: scaler + logistic regression with cross-validation
    clf = Pipeline([
        ('scaler', StandardScaler()),
        ('logreg', LogisticRegression(max_iter=1000))
    ])
    cv = StratifiedKFold(n_splits=min(5, max(2, len(y))), shuffle=True, random_state=42)
    scores = cross_val_score(clf, X, y, cv=cv, scoring='roc_auc')
    print(f"Cross-validated ROC AUC: mean={scores.mean():.3f} std={scores.std():.3f}")
    # fit on full data and save
    clf.fit(X, y)
    joblib.dump(clf, out_model_path)
    return clf, scores
```

```python

# ---------------------- Main pipeline ----------------------

def main(data_dir, out_dir, expand_mm=3.0, age_min=18):
    data_dir = Path(data_dir)
    out_dir = Path(out_dir)
    meshes_dir = out_dir / 'meshes'
    masks_dir = out_dir / 'masks'
    avg_dir = out_dir / 'average'
    model_dir = out_dir / 'model'
    for d in (meshes_dir, masks_dir, avg_dir, model_dir):
        d.mkdir(parents=True, exist_ok=True)

    clin = pd.read_csv(data_dir / 'hvsmr_clinical.csv')
    tech = pd.read_csv(data_dir / 'hvsmr_technical.csv')
    zip_path = data_dir / 'cropped.zip'

    adults = clin[clin['Age'] >= age_min].copy()
    adults['pat_str'] = adults['Pat'].apply(lambda x: f"pat{int(x)}")

    cohort_masks = []
    cohort_ids = []
    spacing_ref = None
    summary_rows = []

    with zipfile.ZipFile(zip_path, 'r') as zf:
        names = zf.namelist()
        for _, row in adults.iterrows():
            pid = row['pat_str']
            img_name, mask_name = find_image_and_mask_in_zip(zip_path, pid)
            if img_name is None:
                print(f"No image found for {pid}. Skipping.")
                continue
            # Load image (not used further here but could be)
            img_arr, spacing = load_nifti_from_zip(zip_path, img_name)
            if spacing_ref is None:
                spacing_ref = spacing
            # Load mask if available
            if mask_name is None:
                print(f"No mask found for {pid}. Skipping segmentation for this patient.")
                continue
            mask_arr, mask_spacing = load_nifti_from_zip(zip_path, mask_name)
            # convert to binary whole-heart
            bin_mask = combine_labels_to_binary(mask_arr, keep_labels=None)
            # expand
            expanded = expand_mask_by_mm(bin_mask, spacing, expand_mm)
            # save mask nifti
            affine = np.diag([spacing[2], spacing[1], spacing[0], 1.0])
            out_mask_path = masks_dir / f"{pid}_expanded.nii.gz"
            nib.save(nib.Nifti1Image(expanded.astype(np.uint8), affine), str(out_mask_path))
            # mesh
            out_mesh = meshes_dir / f"{pid}_expanded.stl"
            mesh_from_mask(expanded, spacing, str(out_mesh))
            # features
            feats = extract_geometric_features(expanded, spacing)
            # label CHD: assume column 'Normal' has 'X' for normal
            is_normal = False
            if 'Normal' in row.index and str(row['Normal']).strip().upper() == 'X':
                is_normal = True
            chd_label = 0 if is_normal else 1
            summary = {'patient': pid, 'age': float(row['Age']), 'chd_label': chd_label}
            summary.update(feats)
            summary_rows.append(summary)
            cohort_masks.append(expanded)
            cohort_ids.append(pid)
            print(f"Processed {pid}: volume={feats['volume_mm3']:.1f} mm3, label={chd_label}")

    # Save summary
    df_summary = pd.DataFrame(summary_rows)
    df_summary.to_csv(out_dir / 'summary.csv', index=False)

    # Build average model if we have masks
    if cohort_masks:
        print("Building average model...")
        avg_prob_path = avg_dir / 'average_prob.nii.gz'
        avg_mesh_path = avg_dir / 'average_mesh.stl'
        mean_prob, avg_mask = build_average_model(cohort_masks, spacing_ref, str(avg_prob_path), str(avg_mesh_path), thresh=0.5)
        print(f"Average model saved to {avg_prob_path} and {avg_mesh_path}")

    # Train CHD classifier if enough samples
    if len(df_summary) >= 4:
        X = df_summary[['volume_mm3', 'surface_area_mm2', 'compactness', 'bbox_x_mm', 'bbox_y_mm', 'bbox_z_mm']].fillna(0).values
        y = df_summary['chd_label'].values
        model_path = model_dir / 'chd_classifier.joblib'
        clf, cv_scores = train_chd_classifier(X, y, str(model_path))
        print(f"Saved classifier to {model_path}")
    else:
        print("Not enough samples to train CHD classifier (need >=4). Saved summary only.")

    print("Pipeline complete.")
```

```python
# ---------------------- CLI ----------------------
if __name__ == '__main__':
    parser = argparse.ArgumentParser()
    parser.add_argument('--data_dir', type=str, required=True, help='Directory containing hvsmr_clinical.csv, hvsmr_technical.csv, cropped.zip')
    parser.add_argument('--out_dir', type=str, required=True)
    parser.add_argument('--expand_mm', type=float, default=3.0)
    parser.add_argument('--age_min', type=float, default=18.0)
    args = parser.parse_args()
    main(args.data_dir, args.out_dir, expand_mm=args.expand_mm, age_min=args.age_min)
```
