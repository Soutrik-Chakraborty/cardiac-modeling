```python
CLINICAL_FILE = '/content/hvsmr_clinical.csv'
TECHNICAL_FILE = '/content/hvsmr_technical.csv'
IMG_FILE = '/content/Pat3/Pat3/pat3_cropped.nii.gz'
SEG_GT_FILE = '/content/Pat3/Pat3/pat3_cropped_seg.nii.gz'
SEG_ENDPOINTS_FILE = '/content/Pat3/Pat3/pat3_cropped_seg_endpoints.nii.gz'
PATIENT_ID = 3
TARGET_LABEL = 1
```

```python
# --- 1. Data Loading and Preprocessing ---

def load_data():
    """Loads CSV data and NIfTI volumes for Patient 3."""
    print("--- 1. Data Loading ---")
    try:
        # Load clinical and technical data
        df_clinical = pd.read_csv(CLINICAL_FILE)
        df_technical = pd.read_csv(TECHNICAL_FILE)

        patient_clinical = df_clinical[df_clinical['Pat'] == PATIENT_ID].squeeze()
        patient_technical = df_technical[df_technical['Pat'] == PATIENT_ID].squeeze()

        print(f"Loaded Clinical Data for Patient {PATIENT_ID}: Category={patient_clinical.get('Category')}, Age={patient_clinical.get('Age')}")
        print(f"Loaded Technical Data for Patient {PATIENT_ID}: TE={patient_technical.get('TE')}ms, TR={patient_technical.get('TR')}ms")

        # Load MRI image and segmentation mask
        img_nii = nib.load(IMG_FILE)
        seg_nii = nib.load(SEG_GT_FILE)

        img_data = img_nii.get_fdata()
        seg_data = seg_nii.get_fdata()

        # Get voxel size (spacing) from the NIfTI header for volume calculation
        voxel_spacing = img_nii.header.get_zooms()[:3]

        print(f"Image shape: {img_data.shape}, Voxel Spacing (mm): {voxel_spacing}")

        # The data is 4D (x, y, z, time/phase). Select the first time phase for static segmentation.
        if img_data.ndim == 4:
            img_data = img_data[..., 0]
            print(f"Reduced image to 3D shape: {img_data.shape}")

        return img_data, seg_data, voxel_spacing, patient_clinical, patient_technical

    except FileNotFoundError as e:
        print(f"Error: One of the expected files was not found: {e}")
        return None, None, None, None, None
    except Exception as e:
        print(f"An unexpected error occurred during loading: {e}")
        return None, None, None, None, None
```

```python
# --- 2. Evaluation Metrics ---

def calculate_dice(pred, gt):
    """Calculates the Dice Similarity Coefficient (DSC)."""
    intersection = np.sum(pred * gt)
    union = np.sum(pred) + np.sum(gt)
    return (2. * intersection) / union if union > 0 else 0.0

def calculate_jaccard(pred, gt):
    """Calculates the Jaccard Index (IoU)."""
    intersection = np.sum(pred * gt)
    union = np.sum(pred) + np.sum(gt) - intersection
    return intersection / union if union > 0 else 0.0

def calculate_hausdorff(pred, gt, spacing=(1.0, 1.0, 1.0)):
    """Calculates the 95th percentile Hausdorff Distance (HD95)."""

    # Check if either mask is empty before proceeding
    if np.sum(pred) == 0 or np.sum(gt) == 0:
        return np.nan # Cannot calculate HD if a mask is empty

    # Get the coordinates of the surface points (boundaries)
    # The mask contains the full volume, we need the surface voxels.
    # An efficient approximation uses the signed distance map (SDM) boundary.
    # For simplicity and robustness in this mock environment, we use non-zero points.

    # Extract coordinates of foreground points
    points_pred = np.argwhere(pred)
    points_gt = np.argwhere(gt)

    # Apply spacing (voxel size) to convert voxel indices to physical distance (mm)
    points_pred_physical = points_pred * spacing
    points_gt_physical = points_gt * spacing

    # Calculate directed Hausdorff distances (A->B and B->A)
    # We use cdist (from scipy.spatial) for efficiency, followed by numpy max/percentile.

    # Note: Scipy's directed_hausdorff is simpler but computationally intensive for large sets.
    # We use a robust approximation:

    # D(A, B): max_{a in A} min_{b in B} ||a - b||
    dist_ab = cdist(points_pred_physical, points_gt_physical).min(axis=1)
    dist_ba = cdist(points_gt_physical, points_pred_physical).min(axis=1)

    # HD95: The 95th percentile of the maximum of the two directed distance sets
    hd95_a_to_b = np.percentile(dist_ab, 95)
    hd95_b_to_a = np.percentile(dist_ba, 95)

    # Symmetric HD95
    symmetric_hd95 = max(hd95_a_to_b, hd95_b_to_a)
    return symmetric_hd95

def evaluate_segmentation(pred_mask, gt_mask, name, spacing):
    """Calculates and returns a dictionary of evaluation metrics."""
    # Ensure masks are binary (0 or 1)
    pred_mask = (pred_mask > 0).astype(np.uint8)
    gt_mask = (gt_mask > 0).astype(np.uint8)

    dice = calculate_dice(pred_mask, gt_mask)
    jaccard = calculate_jaccard(pred_mask, gt_mask)
    hd95 = calculate_hausdorff(pred_mask, gt_mask, spacing)

    # Loss function (Mock: Using 1 - Dice as a simple loss proxy for comparison)
    loss_proxy = 1.0 - dice

    metrics = {
        'Method': name,
        'Dice Score': dice,
        'Jaccard Score': jaccard,
        'Hausdorff Distance (HD95, mm)': hd95,
        'Loss Proxy (1-Dice)': loss_proxy
    }
    return metrics
```

```python
# --- 3. Segmentation Methods ---

# A. Supervised Learning (Deep Learning Mock)

def mock_unet_segmentation(image_data, gt_data):
    """
    MOCK function for Supervised Deep Learning Segmentation (e.g., U-Net).
    In a real scenario, this would involve loading a pre-trained model
    (like a 3D U-Net) and running inference on the image_data.

    For demonstration, we return the ground truth mask itself (perfect prediction)
    to represent the 'best case scenario' of a well-trained supervised model.
    """
    print("Supervised Learning: Mocking U-Net prediction (Perfect Score)")
    # The actual deep learning pipeline would involve Keras/PyTorch model setup.
    # segmentation_mask = model.predict(image_data)

    # For demo: use the ground truth as the prediction
    return gt_data > 0

# B. Unsupervised Learning (K-Means Clustering)

def unsupervised_segmentation(image_data):
    """
    Performs segmentation using K-Means clustering on image intensity values.
    This is a common method for initial, unsupervised image analysis.
    """
    print("Unsupervised Learning: K-Means Clustering (K=4)")
    # Flatten the 3D image volume into a 1D feature vector
    pixels = image_data.reshape(-1, 1)

    # Normalize pixels to aid clustering
    pixels_normalized = minmax_scale(pixels)

    # Assume K=4 clusters: background, myocardium, blood pool, and vessels/noise
    kmeans = KMeans(n_clusters=4, n_init=10, random_state=42)
    kmeans.fit(pixels_normalized)
    labels = kmeans.labels_

    # Reshape the labels back into the original 3D structure
    clustered_image = labels.reshape(image_data.shape)

    # Select the cluster with the highest mean intensity as the "blood pool" (heart cavity)
    cluster_means = []
    for i in range(4):
        cluster_means.append(np.mean(image_data[clustered_image == i]))

    blood_pool_cluster = np.argmax(cluster_means)

    # Create binary mask for the segmented structure
    seg_mask = (clustered_image == blood_pool_cluster)
    return seg_mask.astype(np.uint8)

# C. Combined Learning (Unsupervised Prior Refinement Mock)

def combined_segmentation(image_data, unsupervised_mask):
    """
    MOCK function for Combined/Semi-Supervised Segmentation.
    This approach often uses unsupervised methods (like K-Means or active contours)
    to initialize or refine a supervised prediction.

    For demonstration, we apply a morphological operation (e.g., closing)
    to the unsupervised mask, simulating refinement based on structural knowledge.
    """
    print("Combined Learning: Unsupervised Mask Refinement (Morphological Closing)")

    from scipy.ndimage import binary_closing

    # Mock: Apply closing to fill small holes and smooth boundaries
    # This simulates a refinement step (e.g., using a small amount of labeled data
    # or an anatomical prior to clean up the K-Means output).
    refined_mask = binary_closing(unsupervised_mask, iterations=2)

    return refined_mask.astype(np.uint8)
```

```python
# --- 4. Volumetric Expansion & 3D Visualization ---

def calculate_volumetric_expansion(seg_endpoints_data, voxel_spacing):
    """
    Calculates End Diastolic Volume (EDV), End Systolic Volume (ESV),
    and Ejection Fraction (EF) from the endpoints segmentation.

    Assumes the 4th dimension (time) contains ED and ES phases:
    seg_endpoints_data[..., 0] = ED segmentation
    seg_endpoints_data[..., 1] = ES segmentation

    Volume is calculated as: Number of segmented voxels * Voxel Volume
    """
    voxel_volume_mm3 = np.prod(voxel_spacing)

    # Assume index 0 is ED (largest volume) and index 1 is ES (smallest volume)
    if seg_endpoints_data.ndim < 4:
        print("Error: Endpoints segmentation data is not 4D. Cannot calculate EF.")
        return 0, 0, 0.0

    # Extract ED and ES masks for the target label
    ed_mask = (seg_endpoints_data[..., 0] == TARGET_LABEL)
    es_mask = (seg_endpoints_data[..., 1] == TARGET_LABEL)

    edv_voxels = np.sum(ed_mask)
    esv_voxels = np.sum(es_mask)

    edv_ml = edv_voxels * voxel_volume_mm3 / 1000.0 # Convert mm³ to mL (1 mL = 1000 mm³)
    esv_ml = esv_voxels * voxel_volume_mm3 / 1000.0

    ejection_fraction = ((edv_ml - esv_ml) / edv_ml) * 100 if edv_ml > 0 else 0.0

    return edv_ml, esv_ml, ejection_fraction

def visualize_3d_segmentation(seg_mask, title="3D Heart Segmentation"):
    """
    Creates a basic 3D surface plot of the segmented volume using Marching Cubes.
    """
    print(f"Generating 3D visualization: {title}...")

    # 1. Use marching cubes to extract surface mesh (verts, faces, normals, values)
    # The default level (0.5) works for binary data (0/1)
    try:
        verts, faces, _, _ = marching_cubes(seg_mask, level=0.5)
    except Exception as e:
        print(f"Marching Cubes failed. Skipping 3D visualization: {e}")
        return

    # 2. Create the 3D plot
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    # 3. Create a Poly3DCollection object (mesh)
    mesh = Poly3DCollection(verts[faces], alpha=0.5)
    mesh.set_facecolor('#FF6347') # Tomato red for heart
    ax.add_collection3d(mesh)

    # 4. Configure axis limits and labels
    ax.set_xlim(0, seg_mask.shape[0])
    ax.set_ylim(0, seg_mask.shape[1])
    ax.set_zlim(0, seg_mask.shape[2])

    ax.set_xlabel("X-axis (Slice)")
    ax.set_ylabel("Y-axis (Row)")
    ax.set_zlabel("Z-axis (Column)")

    ax.set_title(title)
    plt.tight_layout()
    plt.show()

# --- Main Pipeline Execution ---

def run_pipeline():
    """Executes the entire HVSMR segmentation and analysis pipeline."""

    # 1. Load Data
    img_data, seg_gt, spacing, clinical, technical = load_data()
    if img_data is None:
        return

    # Load endpoints data for EF calculation
    try:
        seg_endpoints_nii = nib.load(SEG_ENDPOINTS_FILE)
        seg_endpoints_data = seg_endpoints_nii.get_fdata()
    except Exception as e:
        print(f"Warning: Could not load endpoints file for EF calculation. {e}")
        seg_endpoints_data = np.zeros(img_data.shape + (2,)) # Placeholder 4D array

    # Ensure ground truth mask is binary (only keep the target label, assumed to be 1)
    seg_gt = (seg_gt == TARGET_LABEL).astype(np.uint8)

    results_table = []

    # 2. Segmentation and Evaluation

    # --- Supervised Learning ---
    seg_supervised = mock_unet_segmentation(img_data, seg_gt)
    metrics_supervised = evaluate_segmentation(seg_supervised, seg_gt, "Supervised (U-Net Mock)", spacing)
    results_table.append(metrics_supervised)

    # --- Unsupervised Learning (K-Means) ---
    seg_unsupervised = unsupervised_segmentation(img_data)
    metrics_unsupervised = evaluate_segmentation(seg_unsupervised, seg_gt, "Unsupervised (K-Means)", spacing)
    results_table.append(metrics_unsupervised)

    # --- Combined/Hybrid Learning ---
    seg_combined = combined_segmentation(img_data, seg_unsupervised)
    metrics_combined = evaluate_segmentation(seg_combined, seg_gt, "Combined (K-Means + Refinement)", spacing)
    results_table.append(metrics_combined)

    # 3. Print Evaluation Metrics
    results_df = pd.DataFrame(results_table)
    print("\n" + "="*80)
    print("                      SEGMENTATION EVALUATION METRICS                    ")
    print("="*80)
    print(results_df.to_markdown(index=False))
    print("="*80)
    print("\nNote: Supervised results are mocked as perfect (using GT) for comparison.")

    # 4. Volumetric Expansion & Ejection Fraction
    edv, esv, ef = calculate_volumetric_expansion(seg_endpoints_data, spacing)

    print("\n" + "="*80)
    print(f"                   CARDIAC VOLUMETRIC EXPANSION (Patient {PATIENT_ID})                  ")
    print(f"Voxel Spacing: {spacing[0]:.2f}x{spacing[1]:.2f}x{spacing[2]:.2f} mm | Target Structure Label: {TARGET_LABEL}")
    print("-" * 80)
    print(f"End Diastolic Volume (EDV): {edv:.2f} mL")
    print(f"End Systolic Volume (ESV): {esv:.2f} mL")
    print(f"Stroke Volume (SV):         {(edv - esv):.2f} mL")
    print(f"Ejection Fraction (EF):     {ef:.2f} %")
    print("="*80)
    # 5. 3D Visualization
    # This line MUST be inside run_pipeline to access seg_combined.
    visualize_3d_segmentation(seg_combined, title=f"Patient {PATIENT_ID} Heart Segmentation (Combined Method)")

    print("\n3D Visualization displayed in a separate window (close it to continue/exit).")
```
