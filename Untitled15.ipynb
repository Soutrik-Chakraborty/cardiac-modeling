```python
import os
import argparse
import zipfile
import tempfile
import shutil
from pathlib import Path
import numpy as np
import pandas as pd
import nibabel as nib
from scipy.ndimage import distance_transform_edt, binary_closing, binary_opening
from scipy.spatial.distance import cdist
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
from skimage import morphology, measure
import trimesh
import joblib
import matplotlib.pyplot as plt
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

# Optional: torch-based models (UNet and Autoencoder)
try:
    import torch
    import torch.nn as nn
    from torch.utils.data import Dataset, DataLoader
    import torch.nn.functional as F
    TORCH_AVAILABLE = True
except Exception:
    TORCH_AVAILABLE = False
```

```python
def unzip_if_needed(data_dir):
    """If cropped.zip found in data_dir, extract to 'extracted' subfolder and return that path; else return data_dir."""
    data_dir = Path(data_dir)
    zip_path = Path("/content/Pat3.zip") # Convert string to Path object
    if zip_path.exists():
        out = Path("/content/extracted") # Ensure output path is also a Path object
        out.mkdir(exist_ok=True)
        with zipfile.ZipFile(zip_path, 'r') as zf:
            zf.extractall(out)
        return out
    return data_dir
```

```python
def find_image_and_mask(data_root, patient_id):
    """
    Look for NIfTI image and mask for patient_id (e.g., 'pat6').
    Returns image_path, mask_path (one or both may be None).
    """
    data_root = Path(data_root) # Ensure data_root is a Path object
    candidates = list(Path(data_root).rglob(f"*{patient_id}*.nii*"))
    img = None
    mask = None
    for p in candidates:
        name = p.name.lower()
        if "seg" in name or "mask" in name or "label" in name:
            mask = p
        else:
            img = p
    return img, mask
```

```python
def load_nifti(path):
    img = nib.load(str(path))
    arr = img.get_fdata()
    # spacing: nibabel returns zooms (x,y,z)
    zooms = img.header.get_zooms()
    if len(zooms) >= 3:
        spacing = (zooms[2], zooms[1], zooms[0])  # convert to (z,y,x)
    else:
        spacing = (1.0,1.0,1.0)
    return arr.astype(np.float32), spacing
```

```python
def save_nifti(arr, spacing, out_path):
    # simple diagonal affine
    affine = np.diag([spacing[2], spacing[1], spacing[0], 1.0])
    nib.save(nib.Nifti1Image(arr.astype(np.float32), affine), str(out_path))
```

```python
def expand_mask_by_mm(mask, spacing, expand_mm):
    inv = ~mask.astype(bool)
    dt = distance_transform_edt(inv, sampling=spacing)
    expanded = (dt <= expand_mm) | mask.astype(bool)
    return expanded.astype(np.uint8)
```

```python
def mesh_from_mask(mask, spacing, out_mesh_path):
    verts, faces, normals, _ = measure.marching_cubes(mask.astype(np.float32), level=0.5, spacing=spacing)
    mesh = trimesh.Trimesh(vertices=verts, faces=faces, vertex_normals=normals, process=False)
    mesh.export(out_mesh_path)
    return out_mesh_path
```

```python
def dice_score(a, b):
    a = a.astype(bool)
    b = b.astype(bool)
    if a.sum() + b.sum() == 0:
        return 1.0
    return 2.0 * (a & b).sum() / (a.sum() + b.sum())

def jaccard_score(a, b):
    a = a.astype(bool)
    b = b.astype(bool)
    inter = (a & b).sum()
    union = (a | b).sum()
    if union == 0:
        return 1.0
    return inter / union

def hausdorff_distance_sym(a, b, spacing):
    # returns symmetric Hausdorff distance in same units as spacing (mm)
    a_pts = np.argwhere(a.astype(bool))
    b_pts = np.argwhere(b.astype(bool))
    if a_pts.size == 0 or b_pts.size == 0:
        return np.nan
    # convert to mm
    a_mm = a_pts * np.array(spacing)
    b_mm = b_pts * np.array(spacing)
    d_ab = cdist(a_mm, b_mm).min(axis=1).max()
    d_ba = cdist(b_mm, a_mm).min(axis=1).max()
    return max(d_ab, d_ba)
```

```python
def normalize_intensity(img, method="zscore"):
    if method == "zscore":
        m = np.mean(img)
        s = np.std(img) + 1e-9
        return (img - m) / s
    elif method == "minmax":
        mn = np.min(img)
        mx = np.max(img)
        return (img - mn) / (mx - mn + 1e-9)
    return img

def crop_or_pad_to_shape(img, target_shape):
    # center crop or pad with zeros to target_shape (z,y,x)
    img = np.asarray(img)
    out = np.zeros(target_shape, dtype=img.dtype)
    sz = img.shape
    # find start indices
    start = [max(0, (sz[i]-target_shape[i])//2) for i in range(3)]
    end = [start[i] + min(target_shape[i], sz[i]) for i in range(3)]
    out_start = [max(0, (target_shape[i]-sz[i])//2) for i in range(3)]
    out_end = [out_start[i] + (end[i]-start[i]) for i in range(3)]
    out[out_start[0]:out_end[0], out_start[1]:out_end[1], out_start[2]:out_end[2]] = img[start[0]:end[0], start[1]:end[1], start[2]:end[2]]
    return out
```

```python
if TORCH_AVAILABLE:
    class UNet3D(nn.Module):
        def __init__(self, in_ch=1, out_ch=1, base_c=16):
            super().__init__()
            def C(in_c, out_c):
                return nn.Sequential(nn.Conv3d(in_c, out_c, 3, padding=1), nn.BatchNorm3d(out_c), nn.ReLU(inplace=True),
                                     nn.Conv3d(out_c, out_c, 3, padding=1), nn.BatchNorm3d(out_c), nn.ReLU(inplace=True))
            self.enc1 = C(in_ch, base_c)
            self.pool = nn.MaxPool3d(2)
            self.enc2 = C(base_c, base_c*2)
            self.enc3 = C(base_c*2, base_c*4)
            self.bottleneck = C(base_c*4, base_c*8)
            self.up3 = nn.ConvTranspose3d(base_c*8, base_c*4, 2, stride=2)
            self.dec3 = C(base_c*8, base_c*4)
            self.up2 = nn.ConvTranspose3d(base_c*4, base_c*2, 2, stride=2)
            self.dec2 = C(base_c*4, base_c*2)
            self.up1 = nn.ConvTranspose3d(base_c*2, base_c, 2, stride=2)
            self.dec1 = C(base_c*2, base_c)
            self.outc = nn.Conv3d(base_c, out_ch, 1)

        def forward(self, x):
            e1 = self.enc1(x)
            e2 = self.enc2(self.pool(e1))
            e3 = self.enc3(self.pool(e2))
            b = self.bottleneck(self.pool(e3))
            d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
            d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
            d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
            out = self.outc(d1)
            return out

    class HVSMRDataset(Dataset):
        def __init__(self, image_list, mask_list=None, transform=None, target_shape=(64,128,128)):
            self.image_list = image_list
            self.mask_list = mask_list
            self.transform = transform
            self.target_shape = target_shape

        def __len__(self):
            return len(self.image_list)

        def __getitem__(self, idx):
            img = np.asarray(self.image_list[idx], dtype=np.float32)
            img = normalize_intensity(img, "zscore")
            img = crop_or_pad_to_shape(img, self.target_shape)
            x = img[np.newaxis, ...]  # (1, z,y,x)
            if self.mask_list is None:
                return torch.from_numpy(x)
            m = np.asarray(self.mask_list[idx], dtype=np.float32)
            m = crop_or_pad_to_shape(m, self.target_shape)
            y = m[np.newaxis, ...]
            return torch.from_numpy(x), torch.from_numpy(y)
```

```python
def train_unet_supervised(train_images, train_masks, val_images, val_masks, spacing, out_dir, epochs=40, batch_size=2, lr=1e-3, device='cuda'):
    if not TORCH_AVAILABLE:
        raise RuntimeError("PyTorch not available. Install torch to run supervised training.")

    device = torch.device(device if torch.cuda.is_available() else 'cpu')
    model = UNet3D(in_ch=1, out_ch=1, base_c=16).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    bce = nn.BCEWithLogitsLoss()

    train_ds = HVSMRDataset(train_images, train_masks, target_shape=(64,128,128))
    val_ds = HVSMRDataset(val_images, val_masks, target_shape=(64,128,128))
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=1, shuffle=False, num_workers=0)

    history = {'train_loss': [], 'val_loss': [], 'val_dice': []}
    for ep in range(epochs):
        model.train()
        epoch_loss = 0.0
        for xb, yb in train_loader:
            xb = xb.to(device)
            yb = yb.to(device)
            logits = model(xb)
            loss_bce = bce(logits, yb)
            # dice loss
            probs = torch.sigmoid(logits)
            inter = (probs * yb).sum()
            dice_loss = 1 - (2. * inter + 1e-9) / (probs.sum() + yb.sum() + 1e-9)
            loss = loss_bce + dice_loss
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()
        history['train_loss'].append(epoch_loss / len(train_loader))

        # validation
        model.eval()
        val_loss = 0.0
        dice_vals = []
        with torch.no_grad():
            for xb, yb in val_loader:
                xb = xb.to(device)
                yb = yb.to(device)
                logits = model(xb)
                loss_bce = bce(logits, yb)
                probs = torch.sigmoid(logits)
                inter = (probs * yb).sum()
                dice_loss = 1 - (2. * inter + 1e-9) / (probs.sum() + yb.sum() + 1e-9)
                loss = loss_bce + dice_loss
                val_loss += loss.item()
                pred_mask = (probs.cpu().numpy() > 0.5).astype(np.uint8)
                true_mask = yb.cpu().numpy().astype(np.uint8)
                # flatten to compute dice per sample
                for p, t in zip(pred_mask, true_mask):
                    dice_vals.append(dice_score(p[0], t[0]))
        val_loss = val_loss / len(val_loader)
        history['val_loss'].append(val_loss)
        history['val_dice'].append(np.mean(dice_vals))
        print(f"Epoch {ep+1}/{epochs} train_loss={history['train_loss'][-1]:.4f} val_loss={val_loss:.4f} val_dice={history['val_dice'][-1]:.4f}")

    # Save model and history
    os.makedirs(out_dir, exist_ok=True)
    torch.save(model.state_dict(), os.path.join(out_dir, "unet3d_final.pth"))
    joblib.dump(history, os.path.join(out_dir, "training_history.joblib"))
    return model, history
```

```python
def unsupervised_kmeans_seg(img, n_clusters=3, morphological_cleanup=True):
    # img: 3D numpy (z,y,x)
    zyx = np.asarray(img)
    # sample features per voxel: intensity and spatial coordinates (z,y,x scaled)
    coords = np.stack(np.meshgrid(np.arange(zyx.shape[0]), np.arange(zyx.shape[1]), np.arange(zyx.shape[2]), indexing='ij'), axis=-1)
    coords = coords.reshape(-1, 3).astype(np.float32)
    intens = zyx.reshape(-1, 1).astype(np.float32)
    # scale coords to match intensity scale
    scaler_coords = StandardScaler()
    coords_s = scaler_coords.fit_transform(coords)
    feats = np.concatenate([intens, coords_s], axis=1)
    km = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)
    labels = km.fit_predict(feats)
    seg = labels.reshape(zyx.shape)
    # choose cluster with highest mean intensity as "blood/heart" proxy
    cluster_means = [zyx[seg==c].mean() if np.any(seg==c) else 0 for c in range(n_clusters)]
    heart_cluster = int(np.argmax(cluster_means))
    mask = (seg == heart_cluster).astype(np.uint8)
    if morphological_cleanup:
        mask = binary_closing(mask, structure=np.ones((3,3,3)))
        mask = binary_opening(mask, structure=np.ones((3,3,3)))
        mask = morphology.remove_small_objects(mask.astype(bool), min_size=1000).astype(np.uint8)
    return mask
```

```python
if TORCH_AVAILABLE:
    class Simple3DAutoencoder(nn.Module):
        def __init__(self, in_ch=1, base=8):
            super().__init__()
            self.enc = nn.Sequential(
                nn.Conv3d(in_ch, base, 3, padding=1), nn.ReLU(),
                nn.MaxPool3d(2),
                nn.Conv3d(base, base*2, 3, padding=1), nn.ReLU(),
                nn.MaxPool3d(2),
                nn.Conv3d(base*2, base*4, 3, padding=1), nn.ReLU()
            )
            self.dec = nn.Sequential(
                nn.ConvTranspose3d(base*4, base*2, 2, stride=2), nn.ReLU(),
                nn.ConvTranspose3d(base*2, base, 2, stride=2), nn.ReLU(),
                nn.Conv3d(base, in_ch, 1)
            )

        def forward(self, x):
            z = self.enc(x)
            recon = self.dec(z)
            return recon, z

    def train_autoencoder(images, out_dir, epochs=20, batch_size=2, lr=1e-3, device='cuda'):
        if not TORCH_AVAILABLE:
            raise RuntimeError("PyTorch not available.")
        device = torch.device(device if torch.cuda.is_available() else 'cpu')
        model = Simple3DAutoencoder(in_ch=1, base=8).to(device)
        optimizer = torch.optim.Adam(model.parameters(), lr=lr)
        dat = HVSMRDataset(images, mask_list=None, target_shape=(64,128,128))
        loader = DataLoader(dat, batch_size=batch_size, shuffle=True, num_workers=0)
        history = []
        for ep in range(epochs):
            model.train()
            ep_loss = 0.0
            for xb in loader:
                xb = xb.to(device)
                recon, _ = model(xb)
                loss = F.mse_loss(recon, xb)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                ep_loss += loss.item()
            history.append(ep_loss/len(loader))
            print(f"AE Epoch {ep+1}/{epochs} loss={history[-1]:.6f}")
        os.makedirs(out_dir, exist_ok=True)
        torch.save(model.state_dict(), os.path.join(out_dir, "autoencoder.pth"))
        joblib.dump(history, os.path.join(out_dir, "ae_history.joblib"))
        return model, history

    def ae_extract_latent_and_cluster(ae_model, images, device='cuda', n_clusters=2):
        device = torch.device(device if torch.cuda.is_available() else 'cpu')
        ae_model.to(device); ae_model.eval()
        latents = []
        for img in images:
            x = normalize_intensity(img)[np.newaxis, np.newaxis, ...].astype(np.float32)
            with torch.no_grad():
                z = ae_model.enc(torch.from_numpy(x).to(device))
                z = z.cpu().numpy().reshape(-1)
                latents.append(z)
        X = np.stack(latents, axis=0)
        km = KMeans(n_clusters=n_clusters, random_state=0, n_init=10).fit(X)
        return km, X
```

```python
def collect_dataset(data_dir, age_min=None):
    data_root = Path(data_dir)
    data_root = unzip_if_needed(data_root) # Ensure data_root remains a Path object
    clinical = pd.read_csv(Path(data_dir) / "hvsmr_clinical.csv")
    tech = pd.read_csv(Path(data_dir) / "hvsmr_technical.csv")
    if age_min is not None:
        clinical = clinical[clinical['Age'] >= age_min].copy()
    patient_list = clinical['Pat'].astype(int).tolist()
    patient_ids = [f"pat{int(p)}" for p in patient_list]
    samples = []
    for pid in patient_ids:
        img_p, mask_p = find_image_and_mask(data_root, pid)
        samples.append({'pid': pid, 'img': img_p, 'mask': mask_p})
    return samples

def evaluate_segmentation(pred_mask, true_mask, spacing):
    d = dice_score(pred_mask, true_mask)
    j = jaccard_score(pred_mask, true_mask)
    hd = hausdorff_distance_sym(pred_mask, true_mask, spacing)
    return {'dice': d, 'jaccard': j, 'hausdorff_mm': hd}
```

```python
def run_pipeline(data_dir, out_dir, mode="supervised", expand_mm=3.0, age_min=18):
    out_dir = Path(out_dir); out_dir.mkdir(parents=True, exist_ok=True)
    samples = collect_dataset(data_dir, age_min=age_min)
    print(f"Found {len(samples)} patients >= {age_min} years.")
    results = []
    processed_masks = []  # for average model
    spacing_ref = None

    # Simple split for supervised: first 70% train, last 30% val (only for those with masks)
    samples_with_masks = [s for s in samples if s['img'] is not None and s['mask'] is not None]
    imgs_all = []
    masks_all = []
    for s in samples_with_masks:
        img, spacing = load_nifti(s['img'])
        mask, _ = load_nifti(s['mask'])
        if spacing_ref is None:
            spacing_ref = spacing
        imgs_all.append(img)
        masks_all.append((mask>0).astype(np.uint8))

    # Per-patient processing (unsupervised or evaluate)
    if mode in ("unsupervised_kmeans", "unsupervised_classic"):
        for s in samples:
            if s['img'] is None:
                print(f"Skipping {s['pid']} no image")
                continue
            img, spacing = load_nifti(s['img'])
            mask_pred = unsupervised_kmeans_seg(img, n_clusters=3)
            # If ground truth exists, evaluate
            if s['mask'] is not None:
                true_mask, _ = load_nifti(s['mask'])
                true_mask = (true_mask > 0).astype(np.uint8)
                metrics = evaluate_segmentation(mask_pred, true_mask, spacing)
            else:
                metrics = {'dice':np.nan,'jaccard':np.nan,'hausdorff_mm':np.nan}
            # expand & mesh
            expanded = expand_mask_by_mm(mask_pred, spacing, expand_mm)
            mesh_path = Path(out_dir)/f"{s['pid']}_unsup_kmeans_expanded.stl"
            mesh_from_mask(expanded, spacing, str(mesh_path))
            results.append({'pid': s['pid'], 'mode': mode, **metrics})
            processed_masks.append(expanded)
            print(f"{s['pid']} unsup_kmeans dice={metrics['dice']:.4f}")
        # average model
    elif mode == "unsupervised_deep":
        if not TORCH_AVAILABLE:
            raise RuntimeError("PyTorch not installed for deep unsupervised.")
        # train AE on all images with padding/cropping
        ae_out = Path(out_dir)/"autoencoder"
        ae_out.mkdir(exist_ok=True)
        model_ae, hist = train_autoencoder(imgs_all, str(ae_out), epochs=20)
        # get latent clustering and then per-case segmentation (here: naive thresholding of recon error or clustering on latent)
        for s_idx, s in enumerate(samples_with_masks):
            img, spacing = load_nifti(s['img'])
            x = normalize_intensity(img)[np.newaxis, np.newaxis, ...].astype(np.float32)
            import torch
            device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
            model_ae.to(device).eval()
            with torch.no_grad():
                z = model_ae.enc(torch.from_numpy(x).to(device))
                recon = model_ae.dec(z).cpu().numpy()[0,0]
            # residual map thresholding
            resid = np.abs(recon - x[0,0])
            mask_pred = (resid < np.percentile(resid, 50)).astype(np.uint8)  # heuristic
            mask_pred = morphology.remove_small_objects(mask_pred.astype(bool), min_size=1000).astype(np.uint8)
            if s['mask'] is not None:
                true_mask, _ = load_nifti(s['mask'])
                metrics = evaluate_segmentation(mask_pred, (true_mask>0).astype(np.uint8), spacing)
            else:
                metrics = {'dice':np.nan,'jaccard':np.nan,'hausdorff_mm':np.nan}
            expanded = expand_mask_by_mm(mask_pred, spacing, expand_mm)
            mesh_path = Path(out_dir)/f"{s['pid']}_unsup_deep_expanded.stl"
            mesh_from_mask(expanded, spacing, str(mesh_path))
            results.append({'pid': s['pid'], 'mode': mode, **metrics})
            processed_masks.append(expanded)
            print(f"{s['pid']} unsup_deep dice={metrics['dice']:.4f}")
    elif mode == "supervised":
        if not TORCH_AVAILABLE:
            raise RuntimeError("PyTorch not installed for supervised training.")
        # prepare split
        n = len(imgs_all)
        if n < 2: # Need at least 2 samples for train/val split
            print(f"Skipping supervised training: Need at least 2 samples with masks, but found {n}.")
            # Still process and evaluate any samples that DO have masks
            for s_idx, s in enumerate(samples_with_masks):
                 img, spacing = load_nifti(s['img'])
                 true_mask = masks_all[s_idx]
                 # Can't train, so no pred_mask. Metrics will be NaN
                 metrics = {'dice':np.nan,'jaccard':np.nan,'hausdorff_mm':np.nan}
                 # Still attempt to expand and mesh the true mask if needed for visualization
                 expanded = expand_mask_by_mm(true_mask, spacing, expand_mm)
                 mesh_path = Path(out_dir)/f"{samples_with_masks[s_idx]['pid']}_true_expanded.stl"
                 mesh_from_mask(expanded, spacing, str(mesh_path))
                 results.append({'pid': samples_with_masks[s_idx]['pid'], 'mode': mode, **metrics})
                 processed_masks.append(expanded)
                 print(f"{samples_with_masks[s_idx]['pid']} supervised (no training) metrics=NaN")
        else:
            split_idx = int(n * 0.7)
            train_imgs = imgs_all[:split_idx]
            train_masks = masks_all[:split_idx]
            val_imgs = imgs_all[split_idx:]
            val_masks = masks_all[split_idx:]
            unet_out = Path(out_dir)/"unet"
            model, history = train_unet_supervised(train_imgs, train_masks, val_imgs, val_masks, spacing_ref, str(unet_out), epochs=40)
            # inference for all samples_with_masks
            device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
            model.to(device); model.eval()
            for s_idx, s in enumerate(samples_with_masks):
                img, spacing = load_nifti(s['img'])
                x = normalize_intensity(img)[np.newaxis, np.newaxis, ...].astype(np.float32)
                import torch
                with torch.no_grad():
                    logits = model(torch.from_numpy(x).to(device))
                    probs = torch.sigmoid(logits).cpu().numpy()[0,0]
                pred_mask = (probs > 0.5).astype(np.uint8)
                true_mask = masks_all[s_idx]
                metrics = evaluate_segmentation(pred_mask, true_mask, spacing)
                expanded = expand_mask_by_mm(pred_mask, spacing, expand_mm)
                mesh_path = Path(out_dir)/f"{samples_with_masks[s_idx]['pid']}_supervised_expanded.stl"
                mesh_from_mask(expanded, spacing, str(mesh_path))
                results.append({'pid': samples_with_masks[s_idx]['pid'], 'mode': mode, **metrics})
                processed_masks.append(expanded)
                print(f"{samples_with_masks[s_idx]['pid']} supervised dice={metrics['dice']:.4f}")
    elif mode == "hybrid":
        # hybrid: generate pseudo labels via KMeans, pretrain UNet on pseudo labels, then fine-tune on true masks
        if not TORCH_AVAILABLE:
            raise RuntimeError("PyTorch required for hybrid.")
        # generate pseudo labels for all images
        pseudo_masks = []
        for s in samples:
            if s['img'] is None:
                pseudo_masks.append(None); continue
            img, spacing = load_nifti(s['img'])
            pm = unsupervised_kmeans_seg(img, n_clusters=3)
            pseudo_masks.append(pm)
        # Use cases that have true masks: use pseudo for pretraining and true for finetune
        # Build lists
        train_imgs = []
        train_masks = []
        val_imgs = []
        val_masks = []
        for s, pm in zip(samples, pseudo_masks):
            if s['img'] is None: continue
            img, spacing = load_nifti(s['img'])
            if s['mask'] is None:
                # only pseudo
                train_imgs.append(img); train_masks.append(pm)
            else:
                # has real mask -> keep for finetune
                val_imgs.append(img); val_masks.append((load_nifti(s['mask'])[0]>0).astype(np.uint8))
        # Pretrain UNet with pseudo labels
        unet_out = Path(out_dir)/"hybrid"
        model, history = train_unet_supervised(train_imgs, train_masks, val_imgs, val_masks, spacing_ref, str(unet_out), epochs=20)
        # Fine-tune on true masks (if available)
        if len(val_imgs)>0:
            model_ft, history_ft = train_unet_supervised(val_imgs, val_masks, val_imgs, val_masks, spacing_ref, str(unet_out), epochs=10)
            model = model_ft
        # inference and evaluation
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        model.to(device); model.eval()
        for s_idx, s in enumerate(samples):
            if s['img'] is None: continue
            img, spacing = load_nifti(s['img'])
            x = normalize_intensity(img)[np.newaxis, np.newaxis, ...].astype(np.float32)
            import torch
            with torch.no_grad():
                logits = model(torch.from_numpy(x).to(device))
                probs = torch.sigmoid(logits).cpu().numpy()[0,0]
            pred_mask = (probs > 0.5).astype(np.uint8)
            metrics = {'dice':np.nan,'jaccard':np.nan,'hausdorff_mm':np.nan}
            if s['mask'] is not None:
                true_mask = (load_nifti(s['mask'])[0]>0).astype(np.uint8)
                metrics = evaluate_segmentation(pred_mask, true_mask, spacing)
            expanded = expand_mask_by_mm(pred_mask, spacing, expand_mm)
            mesh_from_mask(expanded, spacing, Path(out_dir)/f"{s['pid']}_hybrid_expanded.stl")
            results.append({'pid': s['pid'], 'mode': mode, **metrics})
            processed_masks.append(expanded)
            print(f"{s['pid']} hybrid dice={metrics.get('dice', np.nan):.4f}")
    else:
        raise ValueError("Unknown mode")

    # Save results table
    df_res = pd.DataFrame(results)
    df_res.to_csv(Path(out_dir)/f"results_{mode}.csv", index=False)
    # Build average volumetric model across processed_masks
    if len(processed_masks) > 0 and spacing_ref is not None:
        stack = np.stack(processed_masks, axis=0).astype(np.float32)
        avg_prob = np.mean(stack, axis=0)
        avg_bin = (avg_prob >= 0.5).astype(np.uint8)
        save_nifti(avg_prob, spacing_ref, Path(out_dir)/f"average_prob_{mode}.nii.gz")
        save_nifti(avg_bin, spacing_ref, Path(out_dir)/f"average_bin_{mode}.nii.gz")
        mesh_from_mask(avg_bin, spacing_ref, Path(out_dir)/f"average_mesh_{mode}.stl")
        print("Saved average model and mesh.")
    elif len(processed_masks) > 0 and spacing_ref is None:
        print("Skipping saving average model and mesh: spacing_ref not available.")
    else:
        print("No processed masks to build average model.")
    print("Done.")
```
